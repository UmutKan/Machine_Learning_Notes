```python
#Dependent & Independent
X = df.drop("Outcome", axis=1)
y = df["Outcome"]
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2 ,random_state=15)
```

```python
columns_to_fill= ["Glucose","BloodPressure","SkinThickness", "Insulin", "BMI"]

medians = {}
for col in columns_to_fill:
    medians_values = X_train[X_train[col] != 0][col].median()  #median values for all columns in X
    medians[col] = medians_values
    X_train[col] = X_train[col].replace(0, medians_values)

for col in columns_to_fill:
    X_test[col] = X_test[col].replace(0, medians[col])
```

```python
X_train.describe()
```

```python
#TREE ALGORITMS DOES NOT REQUIRE SCALER PROCESS yet i'm gonna apply it anyways
```

**MODEL TRAINING**

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
from sklearn.ensemble import AdaBoostClassifier
from sklearn.metrics import classification_report, accuracy_score, confusion_matrix
```

```python
ada = AdaBoostClassifier()
ada.fit(X_train, y_train)

y_pred = ada.predict(X_test)
```

```python
print(classification_report(y_test,y_pred))
print(accuracy_score(y_test,y_pred))
print(confusion_matrix(y_test,y_pred))
```

***HYPERPARAMETER TUNING***

```python
from sklearn.model_selection import GridSearchCV
```

```python
adaboost_param = {
    "n_estimators" : [50 ,70, 100, 120, 150, 200],
    "learning_rate" : [0.001, 0.01, 0.1, 1, 10]
}

grid = GridSearchCV(estimator= ada, param_grid= adaboost_param, cv=3, verbose= 1, n_jobs= -1)
grid.fit(X_train, y_train)
```

```python
grid.best_params_
```

```python
ada = AdaBoostClassifier(n_estimators=200, learning_rate=1)

ada.fit(X_train, y_train)
y_pred = ada.predict(X_test)

print(classification_report(y_test,y_pred))
print(accuracy_score(y_test,y_pred))
print(confusion_matrix(y_test,y_pred))
```
